# **Get the date range**

```python
orders_df = spark.table("LH_Olist_Silver.dbo.silver_orders")

date_range = orders_df.select(
    F.min("order_purchase_timestamp").alias("min_date"),
    F.max("order_purchase_timestamp").alias("max_date")
)

date_range.show()
```

# **Create the date dimension**

```python
min_date = orders_df.select(
    F.to_date(F.min("order_purchase_timestamp"))
).first()[0]

max_date = orders_df.select(
    F.to_date(F.max("order_purchase_timestamp"))
).first()[0]

print("Start date:", min_date)
print("End date:", max_date)
```

```python
date_df = (
    spark.sql(
        f"""
        SELECT explode(
            sequence(
                to_date('{min_date}'),
                to_date('{max_date}'),
                interval 1 day
            )
        ) AS date
        """
    )
    .withColumn("date_key", F.date_format("date", "yyyyMMdd").cast("int"))
    .withColumn("year", F.year("date"))
    .withColumn("quarter", F.quarter("date"))
    .withColumn("month", F.month("date"))
    .withColumn("month_name", F.date_format("date", "MMMM"))
    .withColumn("week_of_year", F.weekofyear("date"))
    .withColumn("day_of_month", F.dayofmonth("date"))
    .withColumn("day_of_week", F.dayofweek("date"))
    .withColumn("day_name", F.date_format("date", "EEEE"))
    .withColumn(
        "is_weekend",
        F.when(F.dayofweek("date").isin([1, 7]), True).otherwise(False)
    )
    .select(
        "date_key",
        "date",
        "year",
        "quarter",
        "month",
        "month_name",
        "week_of_year",
        "day_of_month",
        "day_of_week",
        "day_name",
        "is_weekend"
    )
)

print("dim_date rows:", date_df.count())
display(date_df.limit(10))
```

## save to gold

```python
date_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("dim_date")
```

# **# ** fact_order_items****

### Load the required tables

```python
order_items = spark.table("LH_Olist_Silver.dbo.silver_order_items")
orders = spark.table("LH_Olist_Silver.dbo.silver_orders")

dim_customer = spark.table("dim_customer")
dim_product = spark.table("dim_product")
dim_seller = spark.table("dim_seller")
dim_category = spark.table("dim_category")
dim_date = spark.table("dim_date")

print("Order items:", order_items.count())
print("Orders:", orders.count())
```

## Build the fact table

```python
fact_order_items_df = (
    order_items.alias("oi")

    .join(
        orders.alias("o"),
        F.col("oi.order_id") == F.col("o.order_id"),
        "left"
    )

    .join(
        dim_customer.alias("dc"),
        F.col("o.customer_id") == F.col("dc.customer_id"),
        "left"
    )

    .join(
        dim_product.alias("dp"),
        F.col("oi.product_id") == F.col("dp.product_id"),
        "left"
    )

    .join(
        dim_seller.alias("ds"),
        F.col("oi.seller_id") == F.col("ds.seller_id"),
        "left"
    )

    .join(
        dim_category.alias("dcat"),
        F.col("dp.product_category_name") == F.col("dcat.product_category_name"),
        "left"
    )

    .join(
        dim_date.alias("dd"),
        F.to_date(F.col("o.order_purchase_timestamp")) == F.col("dd.date"),
        "left"
    )

    .select(
        F.col("oi.order_id"),
        F.col("oi.order_item_id"),

        F.col("dc.customer_key"),
        F.col("dp.product_key"),
        F.col("ds.seller_key"),
        F.col("dcat.category_key"),
        F.col("dd.date_key"),

        F.col("oi.shipping_limit_date"),

        F.col("o.order_status"),
        F.col("o.order_purchase_timestamp"),
        F.col("o.order_approved_at"),
        F.col("o.order_delivered_carrier_date"),
        F.col("o.order_delivered_customer_date"),
        F.col("o.order_estimated_delivery_date"),

        F.col("oi.price"),
        F.col("oi.freight_value"),

        (
            F.col("oi.price") + F.col("oi.freight_value")
        ).cast("decimal(10,2)").alias("total_item_value")
    )
)

print("fact_order_items rows:", fact_order_items_df.count())

display(fact_order_items_df.limit(10))
```

```python
print("Silver order items:", order_items.count())
print("Gold fact order items:", fact_order_items_df.count())
```

```python
print("Fact rows:", fact_order_items_df.count())

print(
    "Duplicate order-item keys:",
    fact_order_items_df
    .groupBy("order_id", "order_item_id")
    .count()
    .filter(F.col("count") > 1)
    .count()
)
```

```python
fact_order_items_df.select(
    F.count(F.when(F.col("customer_key").isNull(), 1)).alias("missing_customer_key"),
    F.count(F.when(F.col("product_key").isNull(), 1)).alias("missing_product_key"),
    F.count(F.when(F.col("seller_key").isNull(), 1)).alias("missing_seller_key"),
    F.count(F.when(F.col("category_key").isNull(), 1)).alias("missing_category_key"),
    F.count(F.when(F.col("date_key").isNull(), 1)).alias("missing_date_key")
).show()
```

```python
fact_order_items_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("fact_order_items")
```
